## Task 2: VGG16 Feature Extraction Using CIFAR-10

## Task 3: InceptionV3 Transfer Learning

```python
import tensorflow as tf 
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense 

base_model = InceptionV3(weights="imagenet",
                        include_top=False,
                        input_shape=(299, 299, 3))

base_model.trainable = False
for layer in base_model.layers[-30:]:
    layer.trainable = True

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(5, activation="softmax")(x)

model = Model(inputs=base_model.input, outputs=x)

trainable_params = sum(
    tf.keras.backend.count_params(weight)
    for weight in model.trainable_weights
)

print("Trainable parameters:", trainable_params)

model.summary()
```

## Task 4: Fine-Tune ResNet50 on Food Images

```python
import tensorflow as tf
import importlib_resources
import tensorflow_datasets as tfds
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model

dataset, info = tfds.load(
    "food101",
    split="train",
    as_supervised=True,
    with_info=True
)

class_names = info.features["label"].names

class_1 = class_names.index("apple_pie")
class_2 = class_names.index("baby_back_ribs")

print("Class 1:", class_names[class_1])
print("Class 2:", class_names[class_2])

def filter_classes(image, label):
    return tf.logical_or(
        tf.equal(label, class_1),
        tf.equal(label, class_2)
    )

dataset = dataset.filter(filter_classes)

def preprocess_data(image, label):
    image = tf.image.resize(image, (224, 224))
    image = tf.cast(image, tf.float32)
    image = preprocess_input(image)
    label = tf.where(tf.equal(label, class_1), 0, 1)
    return image, label

dataset = dataset.map(
    preprocess_data,
    num_parallel_calls=tf.data.AUTOTUNE
)

dataset = dataset.take(200)
dataset = dataset.shuffle(200)
dataset = dataset.batch(16)
dataset = dataset.repeat()
dataset = dataset.prefetch(tf.data.AUTOTUNE)

base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

for layer in base_model.layers[-10:]:
    layer.trainable = True

x = base_model.output
x = GlobalAveragePooling2D()(x)
output = Dense(2, activation="softmax")(x)

model = Model(
    inputs=base_model.input,
    outputs=output
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    dataset,
    epochs=2,
    steps_per_epoch=13,
    shuffle=False
)

final_accuracy = history.history["accuracy"][-1]

print("Final Training Accuracy:", final_accuracy)
print("Final Training Accuracy (%):", final_accuracy * 100) 
```

## Task 5: InceptionV3 Transfer Lerarning With Dropout 

```python
import tensorflow as tf
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D
from tensorflow.keras.layers import Dense, Dropout

base_model = InceptionV3(
    weights="imagenet",
    include_top=False,
    input_shape=(299, 299, 3)
)


base_model.trainable = False
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.5)(x)


output = Dense(1, activation="sigmoid")(x)
model = Model(inputs=base_model.input, outputs=output)
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])


model.summary()
```
