## Task 1: Prepare Stock Price Sequence Dataset for GRU

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

closing_prices = np.array([
    172.50,174.20,173.80,176.10,178.40,
    177.90,180.20,181.50,179.80,182.30,
    184.10,183.60,185.20,187.40,186.80,
    189.20,191.10,190.50,192.80,194.30,
    193.70,195.60,197.20,196.80,199.10
],dtype=float)

scaler=MinMaxScaler()
scaled_prices = scaler.fit_transform(closing_prices.reshape(-1,1))

X = []
y = []

for i in range(len(scaled_prices) - 10):
    X. append(scaled_prices[i:i + 10])
    y. append(scaled_prices[i + 10])

X = np.array(X)
y = np.array(y)

print("original prices:")
print(closing_prices)
print("X shape:", X.shape)
print("y shape:", y.shape)
print("First Input Sequence:")  
print(X[0]) 
print("First Target:") 
print(y[0]) 
```

## Task 2: Build and Train GRU Model

```python
import numpy as np
import tensorflow as tf  
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

tf.random.set_seed(42)

model = Sequential([GRU(32, input_shape=(10,1)), Dense(1)])
model.compile(optimizer="adam", loss="mse")

history = model.fit(X, y, epochs=10, verbose=1)

print("Final GRU Loss:", history.history["loss"][-1])
```

## Task 3: Train GRU and LSTM and Record Time and Accuracy

```python
import time
import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, LSTM, Dense

tf.random.set_seed(42)

gru_model = Sequential([GRU(32, input_shape=(10,1)), Dense(1)])
gru_model.compile(optimizer="adam", loss="mse", metrics=["accuracy"])

start_time = time.time()

gru_history = gru_model.fit(X, y, epochs=10, verbose=1)

gru_time = time.time() - start_time
gru_loss = gru_history.history["loss"][-1]
gru_accuracy = gru_history.history["accuracy"][-1]

print()

lstm_model = Sequential([LSTM(32, input_shape=(10,1)), Dense(1)])

lstm_model.compile(optimizer="adam", loss="mse", metrics=["accuracy"])

strat_time = time.time()

lstm_history = lstm_model.fit(X, y, epochs=10, verbose=1)

lstm_time = time.time() - start_time
lstm_loss = lstm_history.history["loss"][-1]
lstm_accuracy = lstm_history.history["accuracy"][-1]

print()

print("GRU Training Time:", gru_time, "seconds")
print("GRU Final Loss:", gru_loss)
print("GRU Final Accuracy:", gru_accuracy)

print()

print("LSTM Training Time:", lstm_time, "seconds")
print("LSTM Final Loss:", lstm_loss)
print("LSTM Final Accuracy:", lstm_accuracy)
```

## Task 4: GRU vs LSTM Comparison Table

```python
import pandas as pd 

comparison = pd.DataFrame({
    "Model": ["GRU", "LSTM"],
    "Training Time (seconds)": [gru_time, lstm_time],
    "Final Accuracy": [gru_accuracy, lstm_accuracy],
    "Final Loss": [gru_loss, lstm_loss],
    "Parameters": [gru_model.count_params(), lstm_model.count_params()]
})

print(comparison)
```
